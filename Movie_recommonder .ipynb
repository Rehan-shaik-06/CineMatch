```python
import pandas as pd
import numpy as np
import ast
import nltk
```

```python
tdata=pd.read_csv(r"C:\Users\Rehan\Desktop\Data Science (Code with Harry)\ML\Recemmender System\tmdb_5000_movies.csv")
tcredits=pd.read_csv(r"C:\Users\Rehan\Desktop\Data Science (Code with Harry)\ML\Recemmender System\tmdb_5000_credits.csv")
```

```python
# Join tcredits & [genres,	keywords	,overview , id]
```

```python
df=tdata[['genres','keywords','overview','id']]
```

```python
tcredits=tcredits.rename(columns={'movie_id':'id'})
```

```python
data=tcredits.merge(df,on='id')
```

```python
data.dropna(inplace=True)
```

```python
def top_3_cast(df):
    l=[]
    c=1
    for i in ast.literal_eval(df):
        if c<=3:
            l.append(i['name'])
            c+=1
    return l
    
def generes(df):
    l=[]
    for i in ast.literal_eval(df) :
        l.append(i['name'])
    return l


def director(df):
    l=[]
    for i in ast.literal_eval(df):
        if i["job"]=="Director":
            l.append(i['name'])
    return l

def keywords(df):
    l=[]
    for i in ast.literal_eval(df):
        l.append(i['name'])
    return l
        
```

```python
data
```

```python
data['cast']=data['cast'].apply(top_3_cast)
data['genres']=data['genres'].apply(generes)
data['crew']=data['crew'].apply(director)
data['keywords']=data['keywords'].apply(keywords)
data['overview']=data['overview'].apply(lambda x:x.split())
```

```python
data['cast']=data['cast'].apply(lambda x:[i.replace(" ","") for i in x])
data['crew']=data['crew'].apply(lambda x:[i.replace(" ","") for i in x])
data['genres']=data['genres'].apply(lambda x:[i.replace(" ","") for i in x])
data['keywords']=data['keywords'].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
data['tags']= data['genres'] + data['keywords'] + data['cast'] + data['crew'] + data['overview']
```

```python
movies=data[['id','title','tags']]
```

```python
movies['tags']=movies['tags'].apply(lambda x:" ".join(x))
```

```python
movies['tags']=movies['tags'].apply(lambda x:x.lower())
```

```python
movies.drop_duplicates(subset='title',keep='first',inplace=True)
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(df):
    l=[]
    for i in df.split():
        l.append(ps.stem(i))
    return  " ".join(l)
```

```python
movies['tags']=movies['tags'].apply(stem)
```

```python
movies
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer(max_features=5000,stop_words="english")
```

```python
vectors = cv.fit_transform(movies['tags']).toarray()
```

```python
vectors
```

```python
cv.get_feature_names_out().tolist()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity=cosine_similarity(vectors)
```

```python
def recommendations(df):
    index=movies[movies['title']==df].index[0]
    distances=similarity[index]
    movies_list=sorted(list(enumerate(distances)),reverse=True,key=lambda x:x[1])[1:6]

    for i in movies_list:
        print(movies.iloc[i[0]].title)
```

```python
recommendations('Batman Begins')
```
